```python
#ライブラリをimport
import os
from datetime import datetime
from dotenv import load_dotenv
import requests
import json
```

```python
#envからAPI_KEYを取得
load_dotenv()

API_KEY = os.getenv("YOUTUBE_API_KEY")
```

```python
#動画のカテゴリーIDを取得
category_url = "https://www.googleapis.com/youtube/v3/videoCategories"
category_params = {
    "part": "snippet",
    "regionCode": "JP",
    "key": API_KEY
}
categories = requests.get(category_url, params=category_params)

#raw_dataに保存
category_data=categories.json()
path2 = "../data/raw/category_data.json"
with open(path2, "w", encoding="utf-8") as f:
    json.dump(category_data, f)
```
