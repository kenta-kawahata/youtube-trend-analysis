```python
#ライブラリをimport
import os
from datetime import datetime
from glob import glob

from dotenv import load_dotenv
import json
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import pandas as pd
import seaborn as sns
from sqlalchemy import create_engine
```

```python
load_dotenv()
data_base = os.getenv("DATABASE_URL")
engine = create_engine(f"{data_base}")
```

```python
path = "../data/raw/category_data.json"
with open(path, "r", encoding="utf-8") as f:
    data = json.load(f)
items = data.get("items", [])
category_dic={}
for item in items:
    category_dic[item.get("id")] = [item.get("snippet", {}).get("title")]
```

```python
videos_info_list=[]
files = glob("../data/raw/popular_videos/*.json")
for file in files:
    with open(file, "r", encoding="utf-8") as f:
        data = json.load(f)
    items=data.get("items", [])
    for item, i in zip(items, range(1,51)):
        snippet = item.get("snippet", {})
        statistics = item.get("statistics", {})
        category_id = snippet.get("categoryId")
        video_info={
            "daytime":datetime.strptime(file[42:57], "%Y%m%d_%H%M%S"),
            "rank":i,
            "title" : snippet.get("title"),
            "channel_id": snippet.get("channelId"),
            "channel_title": snippet.get("channelTitle"),
            "category_title":category_dic[f"{category_id}"],
            "view_count" : statistics.get("viewCount"),
            "like_count" : statistics.get("likeCount"),
            "favorite_count" : statistics.get("favoriteCount"),
            "comment_count" : statistics.get("commentCount")
        }
        videos_info_list.append(video_info)
df=pd.DataFrame(videos_info_list)
#SQLに変換
df.to_sql('popular_videos', engine, if_exists='replace', index=False)
```

```python
df
```

```python
#重複を省いた動画数とチャンネル数を確認
print(f"集計時間は{len(df["daytime"].unique())}時間です。")
print(f"データ数は{len(df["title"])}個です。")
print(f"動画数は{len(df["title"].unique())}個です。")
print(f"チャンネル数は{len(df["channel_id"].unique())}個です。")
```

```python
#総データに占める各カテゴリの動画数を確認
count_per_category = pd.read_sql_query(
    '''
    SELECT category_title, COUNT(*) AS category_count
    FROM popular_videos
    GROUP BY category_title
    ORDER BY category_count
    ''',engine)
plt.figure(figsize=(10,5))
sns.barplot(count_per_category, x="category_title", y="category_count")
plt.title("Video count by category")
plt.ylabel("count")
plt.xticks(rotation=45)
plt.show()
```

```python
#時系列によるcategoryの推移を確認
cat_count_hour = pd.read_sql_query(
    '''
    SELECT daytime, category_title, COUNT(category_title) AS category_count
    FROM popular_videos
    GROUP BY daytime, category_title
    ORDER BY daytime
    ''',engine)
plt.figure(figsize=(10,5))
sns.lineplot(data=cat_count_hour, x="daytime", y="category_count", hue="category_title")
plt.title("Category counts every hour")
plt.xticks(rotation=90)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0, fontsize=12)
plt.gca().xaxis.set_major_locator(mdates.HourLocator(interval=6))
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter("%m-%d %H"))
plt.show()
```

```python
#動画のランキング滞在時間のヒストグラムを表示
video_dwell_time = pd.read_sql_query(
    '''
    SELECT title, category_title,COUNT(title) AS title_count
    FROM popular_videos
    GROUP BY title, category_title
    ORDER BY title_count DESC
    ''',engine)
plt.figure(figsize=(10,5))
sns.histplot(video_dwell_time, x="title_count", hue="category_title", multiple="stack", bins=25)
plt.title("Video dwell time on ranking")
plt.xlabel("dwell time(hour)")
plt.ylabel("video count")
plt.xticks(range(0,50,5))
plt.grid(axis="y", linewidth=0.3)
plt.show()
```

```python
#チャンネルのランキング滞在時間のヒストグラムを表示
channel_dwell_time = pd.read_sql_query(
    '''
    SELECT channel_id, category_title, COUNT(channel_id) AS channel_count
    FROM popular_videos
    GROUP BY channel_id, category_title
    ORDER BY channel_count DESC
    ''',engine)
plt.figure(figsize=(10,5))
sns.histplot(channel_dwell_time, x="channel_count", hue="category_title", multiple="stack", bins=30)
plt.title("Channel dwell time on ranking")
plt.xlabel("dwell time(hour)")
plt.ylabel("channel count")
plt.xticks(range(0,60,5))
plt.yticks(range(0,20,2))
plt.grid(axis="y", linewidth=0.3)
plt.show()
```

カテゴリごとのランキング滞在時間では、動画ごとに見ると、music>entertaimment>gamingだが、channelごとに見ると、gaming>music>enntertaimmentの順
つまりチャンネル単位で見ると
